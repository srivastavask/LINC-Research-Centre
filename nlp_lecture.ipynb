```python
# spacy example
import spacy

def run_nlp_morphology_experiment():
    # 1. Load the pre-trained English pipeline
    # Make sure to run 'python -m spacy download en_core_web_sm' in your terminal first.
    try:
        nlp = spacy.load("en_core_web_sm")
    except IOError:
        print("Error: Please download the model first using: python -m spacy download en_core_web_sm")
        return

    # 2. Define test sentences (Declarative, Interrogative, and Complex)
    test_sentences = [
        "The quick brown foxes are running quickly.",
        "Did she watch the historic movie yesterday?",
        "I wanted to study, but I was feeling incredibly tired."
    ]

    print("="*80)
    print("EXPERIMENT: NLP MORPHOLOGICAL ANALYSIS")
    print("="*80)

    for i, sentence in enumerate(test_sentences, 1):
        print(f"\n[Sentence {i}]: \"{sentence}\"")
        print(f"{'Token':<12} | {'Lemma':<12} | {'POS':<6} | {'Morphological Features'}")
        print("-" * 80)

        # Process the text
        doc = nlp(sentence)

        # Extract morphological attributes
        for token in doc:
            # Clean up the morphology dict string for readable output
            morph_feats = str(token.morph) if str(token.morph) else "None"

            print(f"{token.text:<12} | {token.lemma_:<12} | {token.pos_:<6} | {morph_feats}")
        print("-" * 80)

if __name__ == "__main__":
    run_nlp_morphology_experiment()
```

```python
# the transformer model is better than SpaCy model
from transformers import pipeline

ner = pipeline(
    "token-classification",
    model="ai4bharat/IndicNER",
    aggregation_strategy="simple"
)

text = """
नरेंद्र मोदी का जन्म गुजरात के वडनगर में हुआ था।
वह भारत के प्रधानमंत्री हैं।
भारत की राजधानी नई दिल्ली है।
मुंबई महाराष्ट्र का एक प्रमुख शहर है।
"""

results = ner(text)

# for entity in results:
#     print(entity["word"], entity["entity_group"], entity["score"])

for entity in results:
    print(
        entity["word"],
        entity["entity_group"],
        round(entity["score"], 3)
    )
```

```python
from transformers import pipeline

# Load Hindi/Indic NER model
ner = pipeline(
    "token-classification",
    model="ai4bharat/IndicNER",
    aggregation_strategy="max"   # max, simple, none, average, first
)

# Hindi text
text = """
नरेंद्र मोदी का जन्म गुजरात के वडनगर में हुआ था।
वह भारत के प्रधानमंत्री हैं।
भारत की राजधानी नई दिल्ली है।
मुंबई महाराष्ट्र का एक प्रमुख शहर है।
"""

# Perform NER
results = ner(text)

# Display results
print("Named Entities:")
print("-" * 40)

# for entity in results:
#     print(
#         f"{entity['word']} → "
#         f"{entity['entity_group']} "
#         f"({entity['score']:.2f})"
#     )

for entity in results:
    # Extract original text using character positions
    word = text[entity["start"]:entity["end"]]
    print(f"{word} → {entity['entity_group']}")
```

```python
from transformers import pipeline

ner = pipeline(
    "token-classification",
    model="Sankalp-Bahad/Monolingual-Hindi-NER-Model",
    aggregation_strategy="simple"
)

text = """
नरेंद्र मोदी का जन्म गुजरात के वडनगर में हुआ था।
वह भारत के प्रधानमंत्री हैं।
भारत की राजधानी नई दिल्ली है।
मुंबई महाराष्ट्र का एक प्रमुख शहर है।
"""

results = ner(text)

label_map = {
    "NEP": "PERSON",
    "NEL": "LOCATION",
    "NEO": "ORGANIZATION",
    "NEN": "MISCELLANEOUS"
}

for entity in results:
    label = label_map.get(
        entity["entity_group"],
        entity["entity_group"])
    print(f"{entity['word']} → {label}")
```

```python
docs = [
    'the cat sat',
    'the cat sat in the hat',
    'the cat with the hat',
    'the cat with the hat sat on the mat'
]
```

```python
# from keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.text import Tokenizer
tokenize = Tokenizer()
tokenize.fit_on_texts(docs)

print(tokenize.word_index)
print(tokenize.word_counts)
print(tokenize.document_count)
```

```python
vectors = tokenize.texts_to_matrix(docs, mode='count')
print(vectors)
```

```python
print("Vocabulary size:", len(tokenize.word_index))
print("Matrix shape:", vectors.shape)
```

```python
# nltk
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
texts = [
    'the cat sat',
    'the cat sat in the hat',
    'the cat with the hat',
    'the cat with the hat sat on the mat'
]
# tokens = word_tokenize(texts)
tokens = [word_tokenize(text) for text in texts]
print(tokens)
```

```python
vocabulary = sorted(set(word for doc in tokens for word in doc))

print(vocabulary)
```

```python
vectors = []

for doc in tokens:
    vector = [doc.count(word) for word in vocabulary]
    vectors.append(vector)
print(vectors)
```

```python
for i, vector in enumerate(vectors, start=1):
    print(i, vector)
```
